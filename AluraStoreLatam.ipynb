### Importación de datos



```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
facturacion_tienda = tienda['Precio'].sum()
facturacion_tienda2 = tienda2['Precio'].sum()
facturacion_tienda3 = tienda3['Precio'].sum()
facturacion_tienda4 = tienda4['Precio'].sum()


if (facturacion_tienda < facturacion_tienda2) and (facturacion_tienda < facturacion_tienda3) and (facturacion_tienda < facturacion_tienda4):
  print(f'La tienda con menor facturación es la tienda 1 con un total de {facturacion_tienda}')
elif (facturacion_tienda2 < facturacion_tienda) and (facturacion_tienda2 < facturacion_tienda3) and (facturacion_tienda2 < facturacion_tienda4):
  print(f'La tienda con menor facturación es la tienda 2 con un total de {facturacion_tienda2}')
elif (facturacion_tienda3 < facturacion_tienda) and (facturacion_tienda3 < facturacion_tienda2) and (facturacion_tienda3 < facturacion_tienda4):
  print(f'La tienda con menor facturación es la tienda 3 con un total de {facturacion_tienda3}')
else:
  print(f'La tienda con menor facturación es la tienda 4 con un total de {facturacion_tienda4}')
```

```python
plt.plot(['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4'], [facturacion_tienda, facturacion_tienda2, facturacion_tienda3, facturacion_tienda4])
plt.show()
```

# 2. Ventas por categoría

```python
categorias = tienda['Categoría del Producto'].value_counts().nlargest(3)
categorias2 = tienda2['Categoría del Producto'].value_counts().nlargest(3)
categorias3 = tienda3['Categoría del Producto'].value_counts().nlargest(3)
categorias4 = tienda4['Categoría del Producto'].value_counts().nlargest(3)

print(f'Tienda 1\n{categorias}\n')
print(f'Tienda 2\n{categorias2}\n')
print(f'Tienda 3\n{categorias3}\n')
print(f'Tienda 4\n{categorias4}\n')
```

```python
categorias_vendidas = tienda.groupby('Categoría del Producto')['Categoría del Producto'].count().nlargest(3)
categorias_vendidas2 = tienda2.groupby('Categoría del Producto')['Categoría del Producto'].count().nlargest(3)
categorias_vendidas3 = tienda3.groupby('Categoría del Producto')['Categoría del Producto'].count().nlargest(3)
categorias_vendidas4 = tienda4.groupby('Categoría del Producto')['Categoría del Producto'].count().nlargest(3)


categorias_vendidas.plot.barh()
plt.show()
categorias_vendidas2.plot.barh()
plt.show()
categorias_vendidas3.plot.barh()
plt.show()
categorias_vendidas4.plot.barh()
plt.show()
```

# 3. Calificación promedio de la tienda


```python
valoracion_tienda = round(tienda['Calificación'].mean(), 2)
valoracion_tienda2 = round(tienda2['Calificación'].mean(), 2)
valoracion_tienda3 = round(tienda3['Calificación'].mean(), 2)
valoracion_tienda4 = round(tienda4['Calificación'].mean(), 2)

print(f'Tienda 1: {valoracion_tienda}')
print(f'Tienda 2: {valoracion_tienda2}')
print(f'Tienda 3: {valoracion_tienda3}')
print(f'Tienda 4: {valoracion_tienda4}')
```

```python
plt.scatter(['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4'], [valoracion_tienda, valoracion_tienda2, valoracion_tienda3, valoracion_tienda4])
plt.show()
```

# 4. Productos más y menos vendidos

```python
mas_vendido_tienda = tienda['Producto'].value_counts().nlargest(1)
mas_vendido_tienda2 = tienda2['Producto'].value_counts().nlargest(1)
mas_vendido_tienda3 = tienda3['Producto'].value_counts().nlargest(1)
mas_vendido_tienda4 = tienda4['Producto'].value_counts().nlargest(1)
```

```python
print(f'Producto mas vendido Tienda 1:\n {mas_vendido_tienda}\n')
print(f'Producto mas vendido Tienda 2:\n {mas_vendido_tienda2}\n')
print(f'Producto mas vendido Tienda 3:\n {mas_vendido_tienda3}\n')
print(f'Producto mas vendido Tienda 4:\n {mas_vendido_tienda4}\n')
```

# 5. Envío promedio por tienda

```python
envio_promedio_tienda = round(tienda['Costo de envío'].mean(), 2)
envio_promedio_tienda2 = round(tienda2['Costo de envío'].mean(), 2)
envio_promedio_tienda3 = round(tienda3['Costo de envío'].mean(), 2)
envio_promedio_tienda4 = round(tienda4['Costo de envío'].mean(), 2)

print(f'Tienda 1: {envio_promedio_tienda}')
print(f'Tienda 2: {envio_promedio_tienda2}')
print(f'Tienda 3: {envio_promedio_tienda3}')
print(f'Tienda 4: {envio_promedio_tienda4}')
```

```python
plt.plot([envio_promedio_tienda, envio_promedio_tienda2, envio_promedio_tienda3, envio_promedio_tienda4], ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4'])
plt.show()
```
